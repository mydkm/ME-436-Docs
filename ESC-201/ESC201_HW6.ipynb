```python
from sympy.solvers import solve
from sympy import Symbol
import numpy as np
```

```python
# Problem 4.51

ds = 0.625
d = 4.0
l = 5.5
w = 6.0
Es = 29.0 * 10**6
Ec = 3.0 * 10**6
n = Es / Ec
As = (np.pi / 4)*(ds**2)
sigmac = 1.4 * 10**3
sigmas = 2.0 * 10**4

x = Symbol('x')
y = solve(0.5*l*x**2 - n*As*(d - x), x)
# print(n*As)
print(f"{y[0]:.3e}")
print(f"{y[1]:.3e}")

I = (1 / 3)*l*(y[1]**3) + n*As*(d - y[1])**2
Mc = (I*sigmac) / (y[1])
Ms = (I*sigmas) / (n*(4-y[1]))
print(f"{Mc:.3e}")
print(f"{Ms:.3e}")
M = (12 / l)*Ms
print(f"{M:.3e}")
```
