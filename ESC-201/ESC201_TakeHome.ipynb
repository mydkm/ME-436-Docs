This notebook reproduces the calculations used in the Typst writeup for Questions 1, 2, and 3.

All calculations are performed in SI units unless otherwise noted.


```python
import math

def mm(x):
    return x * 1e-3

def inch(x):
    return x * 0.0254

def GPa(x):
    return x * 1e9

def MPa(x):
    return x * 1e6

def N_to_lbf(x):
    return x / 4.4482216152605

def m_to_mm(x):
    return x * 1000

def rad_to_deg(x):
    return x * 180 / math.pi

def print_result(name, value, unit="", sig=6):
    print(f"{name:<35s} = {value:.{sig}g} {unit}".rstrip())
```

```python
# Loads
F_T = 80.0                         # N

# Geometry
t_p = mm(4.0)                      # plate thickness, m
w_p = mm(44.0)                     # plate width, m
t_w = mm(2.0)                      # spar wall thickness, m
d_o = mm(25.0)                     # spar outer side length, m
L_b = mm(76.0)                     # plate/spar overlap length, m
L_s = mm(400.0)                    # free spar length, m
L_total = L_b + L_s                # total beam length, m

# Material properties
E_CF_0 = GPa(40.0)                 # Pa
E_CF_45 = GPa(10.0)                # Pa
E_ASA = GPa(2.1)                   # Pa
sigma_u_0 = MPa(800.0)             # Pa

print_result("F_T", F_T, "N")
print_result("t_p", t_p, "m")
print_result("w_p", w_p, "m")
print_result("t_w", t_w, "m")
print_result("d_o", d_o, "m")
print_result("L_b", L_b, "m")
print_result("L_s", L_s, "m")
print_result("L_total", L_total, "m")
print_result("E_CF_0", E_CF_0, "Pa")
print_result("sigma_u_0", sigma_u_0, "Pa")
```

```python
# Spar section properties
d_i = d_o - 2*t_w
I_s = (d_o**4 - d_i**4) / 12

# Plate section properties
A_p = w_p * t_p
I_pc = w_p * t_p**3 / 12
d_p = (d_o + t_p) / 2

# Joint section property
I_j = I_s + 2*(I_pc + A_p*d_p**2)

# Flexural rigidities
EI_s = E_CF_0 * I_s
EI_j = E_CF_0 * I_j

print_result("d_i", d_i, "m")
print_result("I_s", I_s, "m^4")
print_result("I_s", I_s / 1e-12, "mm^4")
print_result("A_p", A_p, "m^2")
print_result("I_pc", I_pc, "m^4")
print_result("d_p", d_p, "m")
print_result("I_j", I_j, "m^4")
print_result("I_j", I_j / 1e-12, "mm^4")
print_result("E I_s", EI_s, "N*m^2")
print_result("E I_j", EI_j, "N*m^2")
print_result("I_j / I_s", I_j / I_s)
```

```python
theta_Lb = (F_T * L_b / EI_j) * (L_s + L_b/2)
y_Lb = (F_T * L_b**2 / EI_j) * (L_s/2 + L_b/3)

theta_s_free = F_T * L_s**2 / (2 * EI_s)
y_s_free = F_T * L_s**3 / (3 * EI_s)

theta_tip = theta_Lb + theta_s_free
y_tip = y_Lb + theta_Lb * L_s + y_s_free

print_result("theta(L_b)", theta_Lb, "rad")
print_result("theta(L_b)", rad_to_deg(theta_Lb), "deg")
print_result("y(L_b)", y_Lb, "m")
print_result("y(L_b)", m_to_mm(y_Lb), "mm")
print()
print_result("theta_s_free", theta_s_free, "rad")
print_result("y_s_free", y_s_free, "m")
print()
print_result("theta_tip", theta_tip, "rad")
print_result("theta_tip", rad_to_deg(theta_tip), "deg")
print_result("y_tip", y_tip, "m")
print_result("y_tip", m_to_mm(y_tip), "mm")
```

```python
# Equivalent expanded form for y_tip
y_tip_expanded = (
    F_T/EI_j * (L_s*L_b**2 + L_b**3/3 + L_s**2*L_b)
    + F_T*L_s**3/(3*EI_s)
)

print_result("y_tip from compact expression", y_tip, "m")
print_result("y_tip from expanded expression", y_tip_expanded, "m")
print_result("absolute difference", abs(y_tip - y_tip_expanded), "m")
assert abs(y_tip - y_tip_expanded) < 1e-15
```

```python
# Back-of-the-book checks
y_lower = F_T * L_s**3 / (3 * EI_s)
theta_lower = F_T * L_s**2 / (2 * EI_s)

y_upper = F_T * L_total**3 / (3 * EI_s)
theta_upper = F_T * L_total**2 / (2 * EI_s)

print_result("y_lower", m_to_mm(y_lower), "mm")
print_result("theta_lower", theta_lower, "rad")
print_result("theta_lower", rad_to_deg(theta_lower), "deg")
print()
print_result("y_tip", m_to_mm(y_tip), "mm")
print_result("theta_tip", theta_tip, "rad")
print_result("theta_tip", rad_to_deg(theta_tip), "deg")
print()
print_result("y_upper", m_to_mm(y_upper), "mm")
print_result("theta_upper", theta_upper, "rad")
print_result("theta_upper", rad_to_deg(theta_upper), "deg")

assert y_lower < y_tip < y_upper
assert theta_lower < theta_tip < theta_upper
```

```python
# Joint stress at fixed end
M_0 = F_T * L_total
c_j = d_o/2 + t_p
sigma_j = M_0 * c_j / I_j

# Free spar stress at start of free spar
M_Lb = F_T * L_s
c_s = d_o/2
sigma_s = M_Lb * c_s / I_s

sigma_max = max(sigma_j, sigma_s)
n_fos = sigma_u_0 / sigma_max

print_result("M(0)", M_0, "N*m")
print_result("c_j", c_j, "m")
print_result("sigma_j", sigma_j/1e6, "MPa")
print()
print_result("M(L_b)", M_Lb, "N*m")
print_result("c_s", c_s, "m")
print_result("sigma_s", sigma_s/1e6, "MPa")
print()
print_result("controlling sigma", sigma_max/1e6, "MPa")
print_result("factor of safety", n_fos)
```

```python
t_ASA = mm(5.0)

I_ASA = ((d_o + 2*t_ASA)**4 - d_o**4) / 12

# Rough original-joint stiffness:
# spar contribution + ASA collar contribution + two 45-degree plates
EI_j_orig = (
    E_CF_0 * I_s
    + E_ASA * I_ASA
    + 2 * E_CF_45 * (I_pc + A_p*d_p**2)
)

# Same beam formulas, but use EI_j_orig directly for the joint-region stiffness.
theta_Lb_orig = (F_T * L_b / EI_j_orig) * (L_s + L_b/2)
y_Lb_orig = (F_T * L_b**2 / EI_j_orig) * (L_s/2 + L_b/3)

theta_tip_orig = theta_Lb_orig + theta_s_free
y_tip_orig = y_Lb_orig + theta_Lb_orig * L_s + y_s_free

print_result("t_ASA", t_ASA, "m")
print_result("I_ASA", I_ASA, "m^4")
print_result("EI_j_orig", EI_j_orig, "N*m^2")
print_result("EI_j_new", EI_j, "N*m^2")
print_result("EI_j_orig / EI_j_new", EI_j_orig / EI_j)
print()
print_result("theta_tip_orig", theta_tip_orig, "rad")
print_result("y_tip_orig", m_to_mm(y_tip_orig), "mm")
print_result("y_tip_new", m_to_mm(y_tip), "mm")
print_result("y_tip_orig / y_tip_new", y_tip_orig / y_tip)
```

```python
# Simple equal-share screw shear
num_screws_new = 8
d_screw = mm(3.0)

V_screw_simple = F_T / num_screws_new
A_screw = math.pi * d_screw**2 / 4
tau_screw_simple = V_screw_simple / A_screw

print_result("V_screw_simple", V_screw_simple, "N/screw")
print_result("A_screw", A_screw, "m^2")
print_result("A_screw", A_screw/1e-6, "mm^2")
print_result("tau_screw_simple", tau_screw_simple/1e6, "MPa")
```

```python
# Conservative shear-flow estimate for shear transfer between plates and spar
V = F_T
Q_one_plate = A_p * d_p
q = V * Q_one_plate / I_j                  # N/m for one plate-to-spar interface
F_plate = q * L_b
F_total_transfer = 2 * F_plate
V_screw_flow = F_total_transfer / num_screws_new
tau_screw_flow = V_screw_flow / A_screw

print_result("Q_one_plate", Q_one_plate, "m^3")
print_result("q", q, "N/m")
print_result("q", q/1000, "N/mm")
print_result("F_plate", F_plate, "N")
print_result("F_total_transfer", F_total_transfer, "N")
print_result("V_screw_flow", V_screw_flow, "N/screw")
print_result("tau_screw_flow", tau_screw_flow/1e6, "MPa")
```

```python
# Original socket design screw estimate using shear-flow style method

# Original design has 4 screws instead of 8
num_screws_orig = 4

d_p_orig = d_o / 2 + t_ASA + t_p / 2
q_orig = V * E_CF_45 * A_p * d_p_orig / EI_j_orig
F_plate_orig = q_orig * L_b
F_total_orig = 2 * F_plate_orig
V_screw_orig_flow = F_total_orig / num_screws_orig
tau_screw_orig_flow = V_screw_orig_flow / A_screw

print_result("d_p_orig", d_p_orig, "m")
print_result("d_p_orig", m_to_mm(d_p_orig), "mm")
print_result("EI_j_orig", EI_j_orig, "N*m^2")
print_result("q_orig", q_orig, "N/m")
print_result("q_orig", q_orig / 1000, "N/mm")
print_result("F_plate_orig", F_plate_orig, "N")
print_result("F_total_orig", F_total_orig, "N")
print_result("V_screw_orig_flow", V_screw_orig_flow, "N/screw")
print_result("tau_screw_orig_flow", tau_screw_orig_flow / 1e6, "MPa")
```

```python
# Leg geometry
L_leg = inch(8.0)
D_o_leg = inch(0.5)
t_leg = inch(1.0/16.0)
D_i_leg = D_o_leg - 2*t_leg

# Tube properties
A_leg = math.pi/4 * (D_o_leg**2 - D_i_leg**2)
I_leg = math.pi/64 * (D_o_leg**4 - D_i_leg**4)

# Material
E_leg = E_CF_0
sigma_u_leg = sigma_u_0

print_result("L_leg", L_leg, "m")
print_result("D_o_leg", D_o_leg, "m")
print_result("t_leg", t_leg, "m")
print_result("D_i_leg", D_i_leg, "m")
print_result("A_leg", A_leg, "m^2")
print_result("I_leg", I_leg, "m^4")
```

```python
P_material = sigma_u_leg * A_leg

print_result("P_material", P_material, "N")
print_result("P_material", P_material/1000, "kN")
print_result("P_material", N_to_lbf(P_material), "lbf")
```

```python
K = 2.0
P_cr = math.pi**2 * E_leg * I_leg / ((K * L_leg)**2)
sigma_cr = P_cr / A_leg

print_result("P_cr", P_cr, "N")
print_result("P_cr", P_cr/1000, "kN")
print_result("P_cr", N_to_lbf(P_cr), "lbf")
print_result("sigma_cr", sigma_cr/1e6, "MPa")
print_result("sigma_u", sigma_u_leg/1e6, "MPa")
print_result("P_material / P_cr", P_material / P_cr)
```

```python
W_aircraft = 89.0
num_legs = 4
W_leg = W_aircraft / num_legs
n_buckling_static = P_cr / W_leg

print_result("W_aircraft", W_aircraft, "N")
print_result("W_leg", W_leg, "N/leg")
print_result("static buckling FOS", n_buckling_static)
```

```python
c_leg = D_o_leg / 2
P_bend = sigma_u_leg * I_leg / (L_leg * c_leg)

print_result("c_leg", c_leg, "m")
print_result("P_bend", P_bend, "N")
print_result("P_bend", N_to_lbf(P_bend), "lbf")
```
