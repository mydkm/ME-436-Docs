```python
from sympy.solvers import solve
from sympy import Symbol
import numpy as np
```

```python
# Question 6.3

l1 = 0.020
w1 = 0.080
s = 0.050
w2 = 0.120
V = 300
t = 2*l1

I = (w2**4 - w1**4) / 12
A1 = w2 * l1
y1 = 0.050
Q1 = A1 * y1
q = (2*V) / s
V = (q*I) / Q1
Q = Q1 + (l1**2)*w1
tau = (V*Q) / (I*t)

print(f"l1 = {l1:.3f}")
print(f"w1 = {w1:.3f}")
print(f"s = {s:.3f}")
print(f"w2 = {w2:.3f}")
print(f"V = {V:.3f}")
print(f"t = {t:.3f}")
print(f"I = {I:.6f}")
print(f"A1 = {A1:.3f}")
print(f"y1 = {y1:.3f}")
print(f"Q1 = {Q1:.3f}")
print(f"q = {q:.3f}")
print(f"Q = {Q:.3f}")
print(f"tau = {tau:.3f}")
```

```python
# Question 6.15

sigma = 1750.0
tau = 130.0
l = 16.0
w = 5.0
p = 750.0

P = p*l
R = P / 2
M = (1/4)*l*R*12
hb = np.sqrt((6*M) / (w * sigma))
hm = (3*R) / (2*w*tau)

print(f"P = {P:.3f}")
print(f"R = {R:.3f}")
print(f"M = {M:.3f}")
print(f"hb = {hb:.3f}")
print(f"hm = {hm:.3f}")
```

```python
# Question 6.36

l1 = 0.100
l2 = 0.150
w1 = 0.020
w2 = 0.050
t = 0.100
tau = 350000.0

I = ((2*w1 + l1)*((2*w2 + l2)**3) - l1*(l2**3)) / 12
Q = (l1*w2*(w2+l2)) / 2
V = (I*t*tau) / Q

print(f"I = {I:.6f}")
print(f"Q = {Q:.6f}")
print(f"V = {V:.3f}")
```
